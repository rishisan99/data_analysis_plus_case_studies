````python
import numpy as np
````

Loading data

````python
scores = np.loadtxt("./data/survey.txt", dtype='int')
scores[:5]
````

Type and shape checks

````python
print(type(scores))
print(scores.dtype)
print(scores.shape)
````

Creating Detractors category

````python
detractors = scores[scores <= 6]
detractors[:5]
````

Creating Promoters category

````python
promoters = scores[scores >= 9]
promoters[:5]
````

````python
detractors.shape
````

````python
per_of_detractors = detractors.shape[0]/scores.shape[0]
per_of_promoters = promoters.shape[0]/scores.shape[0]
print(f"Percentage of Promoters = {per_of_promoters * 100:.2f}\n\
Percentage of Detractors = {per_of_detractors * 100:.2f}\n")
````

````python
NPS_score = per_of_promoters - per_of_detractors
print(f"NPS Score = {NPS_score * 100:.2f}\n")
````

<span style = "color: red">Insight</span>
> As per the NPS Score the Project from the Organisation failed to meet the expectations and satify the customers, therefore the organisation and its people need to work the product, make it better to meet the audience expectations
