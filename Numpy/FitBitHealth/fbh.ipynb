<span style = "color: green">
AVG Step counts : Happy vs Sad vs Neutral
</span>

```python
happy_avg_step_count = np.mean(step_counts[mood == "Happy"])
print(f"Average step count when Mood is HAPPY: {happy_avg_step_count:.2f}")

sad_avg_step_count = np.mean(step_counts[mood == "Sad"])
print(f"Average step count when Mood is SAD: {sad_avg_step_count:.2f}")

neutral_avg_step_count = np.mean(step_counts[mood == "Neutral"])
print(f"Average step count when Mood is NEUTRAL: {neutral_avg_step_count:.2f}")
```
