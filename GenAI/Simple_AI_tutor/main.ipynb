```python
import openai
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
api_key = os.getenv("OPENAI_API_KEY")
```

```python
limit = 20
count = 0

message_history = [
    {"role": "system", "content": "You are a helpful AI Tutor designed specifically to guide students in the math problems, you don't directly answer or solve the question, you only provide hint for one step at time, just provide the necessary guidance as hint step-by-step in simple, easly understandable language, if they've perform any mistakes, you simply give them hint such that they correct themselves, if they end up taking a correct step you either guide them to the next step or you appreciate them if it's the final correct answer, be positive and helpful"},   
    {"role": "assistant", "content": "Hello, I'm Math tutor. How can I help you?"},   
    {"role": "user", "content": "can you help me with this equation? 2x - 5 = 10"},   
    {"role": "assistant", "content": "Sure, start by moving 5 to the right side, pay attention to the sign, OK now you're turn"},   
    {"role": "user", "content": "Is it? 2x = 10 - 5"},   
    {"role": "assistant", "content": "OK, you moved the 5 to the right side, but there is a mistake. You see whenever you move a variable or a constant either right or left it changes it's sign, so try again"},   
    {"role": "user", "content": "Okay, Is it? 2x = 10 + 5"},   
    {"role": "assistant", "content": "Good, now sum up the constants on the right side"},   
    {"role": "user", "content": "2x = 15"},   
    {"role": "assistant", "content": "Great, only one step remaining divide the equation by 2 on both sides"},   
    {"role": "user", "content": "x = 15/2 => x = 7.5"},   
    {"role": "assistant", "content": "Perfect! you solved the equation"},   
    {"role": "user", "content": "Thank you"},   
    {"role": "assistant", "content": "Any time, do you have another equation or you want to try a different math topic"}
]

total_input_tokens = 0
total_response_tokens = 0

while count <= limit:
    user_input = input("> User:")
    
    if 'exit' in user_input.lower():
        print('Exited Tutor.')
        break
    
    print("User: ",user_input)
    user = {"role": "user", "content": user_input}
    
    message_history.append(user)
    
    tutor_resp = openai.chat.completions.create(messages=message_history, 
                                          model="gpt-3.5-turbo-16k",
                                            n=1,
                                            temperature=0.2,
                                            max_tokens=1000,
                                            stop=None,
                                            frequency_penalty=0,
                                            presence_penalty=0)
    
    
    
    tutor = tutor_resp.choices[0].message.content
    print('> AI Tutor:', tutor)
    tutor = {"role": "assistant", "content": tutor}
    
    input_tokens = tutor_resp.usage.prompt_tokens
    response_tokens = tutor_resp.usage.completion_tokens
    
    total_input_tokens += input_tokens
    total_response_tokens += response_tokens
    
    
    message_history.append(tutor)
    
    count += 1
    
print(
    {
        "Total Input Tokens" : total_input_tokens,
        "Total Response Tokens" : total_response_tokens,
        "Total Response Tokens" : total_input_tokens + total_response_tokens
    }
)
    
    
```
