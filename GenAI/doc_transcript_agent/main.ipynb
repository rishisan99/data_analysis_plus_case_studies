```python
file_path = './data/ap_transcript_aug_26.txt'

with open(file_path, 'r') as f:
    transcript = " ".join(f.readlines())
    
print(transcript)
```

```python
len(transcript)
```

```python
base_instruction = '''You are a helpful financial analyst assistant.
Your task is to answer questions using ONLY the information provided in the transcript.

Rules:
- Answer the question directly and concisely.
- Use specific numbers, percentages, dates, and facts from the transcript wherever relevant.
- Do not make assumptions or infer information that is not explicitly stated in the transcript.
- Do not confuse different financial metrics such as revenue growth, volume growth, PBT growth, and PBT margin.
- Do not add generic conclusions or opinions unless they are directly supported by the transcript.
- If the transcript does not contain enough information to answer the question accurately, respond "I don't know".
'''

question = "How much has Asian Paints' business grown?"

user_prompt = (
    f"Question: {question}\n\n"
    f"Transcript:\n{transcript[1000:20000]}"
)
```

```python
print(user_prompt)
```

```python
messages = [
    {"role": "system", "content": base_instruction},
    {"role": "user", "content": user_prompt}
]
```

```python
responce = openai.chat.completions.create(
    model="gpt-3.5-turbo-16k",
    messages=messages, 
    n=1,
    temperature=0.2,
    max_tokens=1000,
    stop=None,
    frequency_penalty=0,
    presence_penalty=0)
```

```python
print(responce.choices[0].message.content)
```
